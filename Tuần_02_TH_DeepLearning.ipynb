### Bài 10: Phép toán điều kiện trên phần tử (Element-wise Condition)

```python
x = tf.random.uniform(shape=[5], minval=1, maxval=10, dtype=tf.int32)
y = tf.random.uniform(shape=[5], minval=1, maxval=10, dtype=tf.int32)

# Tạo ma trận điều kiện so sánh phần tử
điều_kiện = x < y

# Nếu điều kiện Đúng chọn (x + y), ngược lại chọn (x - y)
bài_10 = tf.where(điều_kiện, x + y, x - y)

print("Tensor x: ", x.numpy())
print("Tensor y: ", y.numpy())
print("Kết quả : ", bài_10.numpy())
```

### Bài 11: Cộng ma trận và giữ nguyên giá trị đường chéo gốc

```python
# Điều chỉnh kích thước thành (3, 3) để phép toán đường chéo hợp lệ
x = tf.random.uniform(shape=(3, 3), minval=5, maxval=10, dtype=tf.int32)
y = tf.random.uniform(shape=(3, 3), minval=15, maxval=20, dtype=tf.int32)

# Thực hiện phép cộng x + y
tong_xy = x + y

# Lấy các phần tử trên đường chéo chính của x
duong_cheo_x = tf.linalg.diag_part(x)

# Khôi phục đường chéo chính của x vào ma trận tổng
bài_11 = tf.linalg.set_diag(tong_xy, duong_cheo_x)

print("Ma trận x ban đầu:\n", x.numpy())
print("Ma trận tổng x + y ban đầu:\n", tong_xy.numpy())
print("Kết quả sau khi giữ lại đường chéo của x:\n", bài_11.numpy())
```

### Bài 12: Đóng gói các hàm tính toán vào môi trường Graph Execution

```python
# Dùng decorator biên dịch hàm chạy trên Graph của TensorFlow
@tf.function
def tinh_toan_tren_graph(tensor_a, tensor_b):
    cong = tf.add(tensor_a, tensor_b)
    tru = tf.subtract(tensor_a, tensor_b)
    nhan = tf.multiply(tensor_a, tensor_b)
    chia = tf.divide(tensor_a, tensor_b)
    return cong, tru, nhan, chia

# Khởi tạo 2 tensor ngẫu nhiên kích thước (3, 2, 2)
x = tf.random.normal(shape=(3, 2, 2))
y = tf.random.normal(shape=(3, 2, 2))

# Gọi hàm chạy trực tiếp trong Graph
kq_cong, kq_tru, kq_nhan, kq_chia = tinh_toan_tren_graph(x, y)

print("Đã thực thi tính toán Cộng, Trừ, Nhân, Chia trên Graph thành công!")
print("Kết quả phép cộng: ", kq_cong)
```

### Bài 13 & 14: Xây dựng Model tính toán (Tổng, Hiệu, Tích)

```python
import tensorflow as tf

# Bài 13: Model tính tổng
class SumModel(tf.Module):
    def __call__(self, x, y):
        return tf.add(x, y)

# Bài 14: Model tính hiệu và tích
class DiffProdModel(tf.Module):
    def __call__(self, x, y):
        diff = tf.subtract(x, y)
        prod = tf.multiply(x, y)
        return diff, prod

# Kiểm tra:
x = tf.constant([10.0, 20.0])
y = tf.constant([5.0, 5.0])

sum_model = SumModel()
diff_prod_model = DiffProdModel()

print("Tổng:", sum_model(x, y).numpy())
d, p = diff_prod_model(x, y)
print("Hiệu:", d.numpy(), "| Tích:", p.numpy())
```

### Bài 15: Xây dựng Model tính chu vi hình chữ nhật

```python
class RectanglePerimeterModel(tf.Module):
    def __call__(self, a, b):
        return (a + b) * 2

# Kiểm tra:
rect_model = RectanglePerimeterModel()
a = tf.constant(5.0)
b = tf.constant(3.0)
print("Chu vi hình chữ nhật:", rect_model(a, b).numpy())
```

### Bài 16: Xây dựng Model tính diện tích toàn phần hình hộp chữ nhật

```python
class BoxAreaModel(tf.Module):
    def __call__(self, a, b, c):
        s_xung_quanh = (a + b) * c * 2
        s_toan_phan = (a * b) * 2 + s_xung_quanh
        return s_xung_quanh, s_toan_phan

# Kiểm tra:
box_model = BoxAreaModel()
a = tf.constant(3.0)
b = tf.constant(4.0)
c = tf.constant(5.0)
sxq, stp = box_model(a, b, c)

print(f"Diện tích xung quanh: {sxq.numpy()}")
print(f"Diện tích toàn phần: {stp.numpy()}")
```
